# Example pLDDT plot

```python
# ----- LOAD DATA -----
df = pd.read_csv('../data/Q9X078.csv.gz')

# ----- PLOT -----

fig, axs = plt.subplots(
    2,1, figsize=(6,4), sharex=True, 
    subplot_kw = {'projection': 'secstruct'},
    gridspec_kw={'height_ratios': [1, 7], 'hspace': .04}
)

# Main axis
ax = axs[1]
ax.grid()
ax.plot(df['seqIndex'], df['pLDDT'], label='pLDDT')
ax.set_ylabel('pLDDT')
ax.set_xlabel('Seq. index')
ax.legend(loc='upper left', bbox_to_anchor=(1., 1.),)

# Secondary structure axis
ax = axs[0]
ax.axes.set_axis_off()
ax.draw_secondary_structure(
    df['strideLabel'], x=1, height=.15, alpha=.9, artist='stride'
)
ax.legend(
    loc='center left', bbox_to_anchor=(1., .5),
    bbox_transform=axs[1].transAxes # Center-left on main axis
)
```

# Style examples

```python
import matplotlib.pyplot as plt
import secstructartist

HSL_STYLES = [
    'hsl-outlined',
    'hsl-filled',
    'hsl-pymol',
    'hsl-pymol2',
    'hsl-pymol3',
]
HSL_STRING = 'LLLLSSSSSSSSLLLLSSSLLLLLHHHHHHHHHHHHHHHHHHLLLLLSSSSSSSLLLHHLLLLLLSSSHHHHLLLLLLLLL'
HSL_LABELS = ['Helix', 'Sheet', 'Loop']

DSSP_STYLES = [
    'dssp-dssp',
    'dssp-stride'
]
DSSP_STRING = '   S EEEEEEEETPTEEEEEEE  TTHHHHHHHHIIII  BBBB  HHHHHHHHTTSTT GGGG  EEEEEE  PPPPP '
DSSP_LABELS = [
    '$3_{10}$-Helix', '$\\alpha$-Helix', '$\\pi$-Helix',
    'Extended $\\beta$-Strand', '$\\beta$-Bridge', 'H-bonded turn',
    '$\\kappa$-Helix', 'Bend', 'Loop'  
]



fig, axs = plt.subplots(
    2,1, figsize=(4, 8), dpi=120, sharex=True,
    subplot_kw={'projection': 'secstruct'},
    gridspec_kw={'height_ratios': [5,2], 'hspace': .55}
)

ax = axs[0]
# ax.set_title('HSL styles')
ax.axes.set_axis_off()
for i, style in enumerate(HSL_STYLES):
    ax.axhline(.5-i, color='grey', linestyle='--', linewidth=.5)
    ax.text(-3,-i, style, ha='right', va='center')
    ax.draw_secondary_structure(HSL_STRING, y=-i, height=.45, artist=style)

    if style == 'hsl-pymol':
        handles = {
            lbl: hdl for hdl, lbl in zip(*ax._ssa_secstructartist.get_legend_handles_labels())
        }
        handles = [handles[lbl] for lbl in HSL_LABELS]
        ax.legend(
            handles, HSL_LABELS, loc='lower center', bbox_to_anchor=(.4, 1),
            title='HSL styles', title_fontsize=12, fontsize=9, ncols=3, 
        )        
ax.axhline(-.5-i, color='grey', linestyle='--', linewidth=.5)
ax.set_ylim([-.3-i, .3])


ax = axs[1]
# ax.set_title('DSSP/STRIDE styles')
ax.axes.set_axis_off()
for i, style in enumerate(DSSP_STYLES):
    ax.axhline(.5-i, color='grey', linestyle='--', linewidth=.5)
    ax.text(-3,-i, style, ha='right', va='center')
    ax.draw_secondary_structure(DSSP_STRING, y=-i, height=.45, artist=style)

    if style == 'dssp-dssp':
        handles = {
            lbl: hdl for hdl, lbl in zip(*ax._ssa_secstructartist.get_legend_handles_labels())
        }
        handles = [handles[lbl] for lbl in DSSP_LABELS]
        ax.legend(
            handles, DSSP_LABELS, loc='lower center', bbox_to_anchor=(.4, 1),
            title='DSSP/STRIDE styles', title_fontsize=12, fontsize=9, ncols=3, 
        )      
ax.axhline(-.5-i, color='grey', linestyle='--', linewidth=.5)
ax.set_ylim([-.3-i, .3])
```
